**1. Print numbers from 1 to 100 using a loop.**

```python
for i in range(0,101):
  print(i)
```

**3. Find the sum of first n natural numbers.**

```python
sum = 0
num = int(input("Enter the value of n"))
for i in range(0,num+1):
  sum += i
print(sum)
```

**4. Find factorial of a number.**

```python
fact = 1
num = int(input("Enter the number"))
for i in range(1,num+1):
  fact *= i
print(fact)
```

**5. Reverse a number.**

```python
num = int(input("Enter the number"))
rev = 0
while num != 0:
  rem = num % 10
  rev = rev*10 + rem
  num = num // 10
print(rev)
# int division (//) khyal garne
```

**6. Count number of digits in a number.**

```python
num = int(input("Enter the number"))
count = 0
while num != 0 :
  rem = num % 10
  count = count + 1
  num = num // 10
print(count)

```

**7. Find sum of digits.**

```python
num = int(input("Enter the number"))
sum = 0
while num!= 0:
  rem = num % 10
  sum = sum + rem
  num = num // 10
print(sum)
```

**8. Check whether a number is palindrome.**

```python
rev = 0
num = int(input("Enter the number"))
original = num
while num != 0:
  rem = num % 10
  rev = rev*10 + rem
  num = num // 10
if rev == original:
  print("The number is palindrome")
else:
  print("The number is not palindrome")

```

**9. Print multiplication table of a number.**
